```python
import numpy as np 
import pandas as pd 
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
age=[10,20,30,40,50,60,70,80]
data=np.random.randint(1,100,8)
```

```python
plt.plot(age,data,"-*g")
plt.xlabel("AGE")
plt.ylabel("DATA")
plt.title("DATA RESULTS")
plt.show()
```

```python
data2=np.linspace(0,100,8)
data2
```

```python
plt.subplot(1,2,1)
plt.plot(age,data,'r--')
plt.subplot(1,2,2)
plt.plot(age,data2,'b')
```

```python
#my_figure=plt.figure()
#figure_axes=my_figure.add_axes([0.9,0.9,0.5,0.5])
my_figure, figure_axes = plt.subplots()

figure_axes.plot(data2,data2**2,color="#FF00FF",label="x ** 2",linewidth = 3.0)

figure_axes.plot(data2,(data2**2)+1000,color="black",label="x ** 2 + 1000",alpha=0.3)

figure_axes.plot(data2,(data2**2)+2000,color="cyan",label="x ** 2 + 2000",linestyle="-.")

figure_axes.plot(data2,(data2**2)+3000,color="#7B68EE",label="x ** 2 + 3000",marker="o",markerfacecolor="white",linestyle=":")

figure_axes.set_xlabel("X LABEL")
figure_axes.set_ylabel("Y LABEL")
figure_axes.set_title("CUBE OF NUMBER")

figure_axes.legend(loc=2)
```

```python
weight=np.random.randint(1,50,30)
height=np.random.random(30)
```

```python
figure1, axes1 = plt.subplots(nrows=1,ncols=3)


axes1[0].scatter(weight,height,color="red")
axes1[0].set_title("SCATTER")

axes1[1].hist(weight,color="orange")
axes1[1].set_title("HISTOGRAM")

axes1[2].boxplot(height,patch_artist=True, boxprops=dict(facecolor="yellow"))
axes1[2].set_title("BOXPLOT")

plt.tight_layout()
plt.show()
```
