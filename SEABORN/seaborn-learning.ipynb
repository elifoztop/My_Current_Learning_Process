```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
athlete_db=pd.read_csv("/kaggle/input/datasets/heesoo37/120-years-of-olympic-history-athletes-and-results/athlete_events.csv")
athlete_db.head()
```

```python
basketball_data = athlete_db.loc[athlete_db["Sport"] == "Basketball"].copy()
```

```python
fig , axes = plt.subplots(figsize=(10,5),dpi=150)

axes.scatter("Height","Age",data=basketball_data,color="orange")
axes.set_xlabel("Height")
axes.set_ylabel("Age")
axes.set_title("MATPLOTLIB BASKETBALL SUBPLOTS")

plt.tight_layout()
plt.show()
```

```python
fig , axes = plt.subplots(figsize=(10, 5),dpi=150)

sns.set_style("whitegrid")

sns.scatterplot(data=basketball_data, 
                x="Height",y="Age",
                ax=axes,
                hue="Sex",
                palette={"M": "#1e88e5", "F": "#ff69b4"},
                #style="Medal",
                #markers={"Gold":"D","Silver":"s","Bronze":"^"}
                #size="Year"
                )

axes.set_xlabel("Height")
axes.set_ylabel("Age")
axes.set_title("SEABORN BASKETBALL SUBPLOTS")

plt.tight_layout()
plt.show()
```

```python
fig , axes = plt.subplots(figsize=(10, 5),dpi=150)

sns.set_style("whitegrid")

sns.lineplot(data=basketball_data, 
                x="Height",y="Age",
                ax=axes,
                hue="Sex",
                palette={"M": "#1e88e5", "F": "#ff69b4"}
            )

axes.set_xlabel("Height")
axes.set_ylabel("Age")
axes.set_title("SEABORN BASKETBALL LINEPLOT")

plt.tight_layout()
plt.show()
```

```python
fig , axes = plt.subplots(figsize=(10, 5),dpi=150)

sns.set_style("whitegrid")

sns.histplot(data=basketball_data, 
             x="Height",
             ax=axes,
             hue="Sex",
             palette={"M": "#1e88e5", "F": "#ff69b4"}
            )

axes.set_xlabel("Height")
axes.set_ylabel("Density")
axes.set_title("SEABORN BASKETBALL HISTPLOT")

plt.tight_layout()
plt.show()
```

```python
sns.set_style("whitegrid")

sns.displot(
    data=basketball_data, 
    x="Height",
    hue="Sex",
    palette={"M": "#1e88e5", "F": "#ff69b4"},
    height=5,
    aspect=2,
    kind="kde",
)

plt.xlabel("Height")
plt.ylabel("Density")
plt.title("SEABORN BASKETBALL DİSPLOT(kde)")

plt.show()
```

```python
child_data=athlete_db.loc[athlete_db["Age"]<=18].copy()
child_data.describe()
```

```python
plt.figure(figsize=(10, 5), dpi=150)

sns.set_style("whitegrid")

sns.countplot(x="Age",
            data=child_data,
            hue="Sex",
            palette={"M":"#06B6D4","F":"#FBBF24"})

plt.title("Child Athlete Count")
```

```python
medal_data = child_data.loc[child_data["Medal"].notna()].copy()

plt.figure(figsize=(10, 5), dpi=200)

sns.set_style("whitegrid")
sns.countplot(x="Age",
              data=medal_data, 
              hue="Sex",
              palette={"M":"#06B6D4","F":"#FBBF24"}
             )

plt.title("Child Medalists by Age and Sex")
```

```python
plt.figure(figsize=(10,5),dpi=150)

sns.set_style("whitegrid")

sns.boxplot(x="Medal",y="Age",
            data=child_data,
            hue="Sex",
            palette={"M":"#06B6D4","F":"#FBBF24"}
            )

plt.title("Child Athlete: Age Distribution by Medal and Sex")
```

```python
plt.figure(figsize=(10,5),dpi=150)

sns.set_style("whitegrid")

sns.violinplot(x="Medal",y="Year",
               data=child_data,
               hue="Sex",
               palette={"M":"#06B6D4","F":"#FBBF24"}
               )
plt.title("Child Athlete: Year Distribution by Medal and Sex")
```
